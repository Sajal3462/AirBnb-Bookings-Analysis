````python
# Function to get all id mean & median while passing area code and dataframe.

# Function to calculate mean & median for a specific id
def id_stats(listing_id, df):
    """
    Returns mean and median of all numerical columns for a specific listing ID.
    """
    id_df = df[df['id'] == listing_id]
    numeric_data = id_df.select_dtypes(include='number')
    mean_vals = numeric_data.mean().to_frame(name='Mean')
    median_vals = numeric_data.median().to_frame(name='Median')
    return pd.concat([mean_vals, median_vals], axis=1)
````
